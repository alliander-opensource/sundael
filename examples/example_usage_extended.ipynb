# Usage example extended 
The files in `examples/data` are based on data sets from [Liander open data](https://www.liander.nl/over-ons/open-data), 
specifically "Verbruiksdata slimme meter 2012-2014".

This example requires `hvplot` to be installed for plotting.

```python
from pathlib import Path

import hvplot.pandas  # noqa: F401
import pandas as pd

from sundael import PVConfig, PVDisaggregator
from sundael.disaggregation import (
    calculate_rule_based_self_consumption,
    correct_generation_with_net_gen,
)

data_dir = Path("./data")
```

## Basic example with plots of the inbetween steps

```python
#read in the example data and show some samples
pv_ratio = pd.read_parquet(data_dir / "pv_ratio.parquet")
net_con = pd.read_parquet(data_dir / "net_con.parquet").T
net_gen = pd.read_parquet(data_dir / "net_gen.parquet").T
```

```python
pv_ratio.sample(3)
```

```python
net_con.sample(3).iloc[:, :5]
```

```python
net_gen.sample(3).iloc[:, :5]
```

```python
# Plot one example of "Klant 63" import/export measurements

idx = "Klant 63"

pd.concat(
    (
        net_con.loc[idx].rename("Import"),
        net_gen.loc[idx].rename("Export") * -1,
    ),
    axis=1,
).fillna(0).hvplot(width=1000, title=idx)
```

```python
area_mapping = {k: "average" for k in net_con.index} # Note: mapping is optional if only one reference (area) is present
pv = PVDisaggregator(pv_ratio=pv_ratio, area_mapping=area_mapping) #Load module

# calculate the net load from the net_con/net_gen (import/export) smart meter values and plot them
net_con, net_gen, net_load = pv._calculate_load(net_con, net_gen, None)
idx = "Klant 63"
display(
    pd.concat(
        (
            net_con.loc[idx].rename("net consumption"),
            net_gen.loc[idx].rename("net generation") * -1,
            # plot net load times -1 so that sign convention is the same as in the other plots
            net_load.loc[idx].rename("net load") * -1, 
        ),
        axis=1,
    ).hvplot(width=1000, title=f"Input data: {idx}")
)
```

```python
#Fit the optimal installation parameters for all customers and show them in a table
pv._validate_area(net_load) # Check that all areas in net_load are present in pv_ratio
params = pv.optimize_parameters(net_load)
params
```

```python
# Calculate estimated and perform two corrections:
#        * If estimated generation is 0, the net_gen value will be used (if higher than 0).
#        * If the maximum of the net_gen is higher than that of the generation, all
#            generation values will be scaled to maximum(net_gen)/maximum(generation). 
# Notice that after this, the generation could be still lower that the net_gen/export values
estimated_generation = pv.estimate_generation(net_load, params)
corrected_generation = correct_generation_with_net_gen(estimated_generation, net_gen) 
display(
    pd.concat(
        (
            net_gen.loc[idx].rename("Export smart meter data"),
            estimated_generation.loc[idx].rename("estimated generation"),
            corrected_generation.loc[idx].rename("corrected generation"),
        ),
        axis=1,
    ).hvplot(width=1000, title=f"Estimated and corrected generation: {idx}")
)
```

```python
# Calculate two different estimations of the self consumption and take the average to get at the final estimation
rule_based_self_consumption = calculate_rule_based_self_consumption(
    generation=corrected_generation,
    net_con=net_con,
    net_gen=net_gen,
)
simple_self_consumption = pv.calculate_simple_self_consumption(
    generation=corrected_generation,
    net_gen=net_gen,
)
self_consumption = (rule_based_self_consumption + simple_self_consumption) / 2
pv.calculate_self_consumption_mpi(self_consumption)

display(
    pd.concat(
        (
            simple_self_consumption.loc[idx].rename("optimistic self consumption"),
            self_consumption.loc[idx].rename("average estimated self consumption"),
            rule_based_self_consumption.loc[idx].rename("conservative self consumption")
        ),
        axis=1,
    ).hvplot(width=1000, title=f"Different estimated self consumption: {idx}")
)
```

```python
# Show table of model performance indicators. If the self_consumption_mpi is
# much larger then one, serious corrections where needed
pv.mpi[pv.mpi['year'] == 2013]
```

```python
# Calculate gross generation and consumption and put both input and output into one plot
generation = net_gen + self_consumption
consumption = net_con + self_consumption

result = {
    "params": params,
    "net_con": net_con,
    "net_gen": net_gen,
    "generation": generation,
    "consumption": consumption,
}


display(
    pd.concat(
        (
            net_gen.loc[idx].rename("Export smart meter data") * -1,
            net_con.loc[idx].rename("Import smart meter data"),
            generation.loc[idx].rename("generation") * -1,
            consumption.loc[idx].rename("consumption"),
        ),
        axis=1,
    ).hvplot(width=1000, title=f"Final result: {idx}")
)
```

## Examples with custom configuration

The previous example assumes the data originates from the Netherlands. However, the code will
also work for other localities. In this case, you will have to provide location (latitude and
longitude) and the temperature. 

Note that the timezone of the `DateTimeIndex` should be set (it will be converted to UTC internally).

The most straightforward way is to provide a default via the config. In this case, the `auto_infer_temp` 
value should be set to `False`. 

```python
# Example values for the Netherlands
config = PVConfig(latitude=52.3, longitude=4.7, yearly_avg_temp=11.3, auto_infer_temp=False)

pv = PVDisaggregator(pv_ratio=pv_ratio, config=config)

result = pv.disaggregate(net_con, net_gen)
```

You can also set the latitude and longitude specifically for multiple geographical areas. The
following example shows how to do this for a synthetic data set. In reality, you would provide
a PV ratio for each area, with an accompanying latitude and longitude.

```python
# Note: this is fake data. The Liander data set is anonymized and contains no geographic information!

# Latitude and longitude for each area (Dutch cities in this synthetic example)
area_latlon = pd.DataFrame(
    {
        "latitude": [52.3676, 51.9225, 52.0705, 52.0902, 51.4416, 53.2129, 50.8308, 52.3667, 51.5597, 51.5897],
        "longitude": [4.9041, 4.4792, 4.3007, 5.1075, 5.4697, 6.5795, 5.6915, 5.1833, 5.0919, 4.7789],
    },
    index=[
        "Amsterdam",
        "Rotterdam",
        "Den Haag",
        "Utrecht",
        "Eindhoven",
        "Groningen",
        "Maastricht",
        "Almere",
        "Tilburg",
        "Breda",
    ],
)

# A separate PV ratio for each area.
pv_ratio_area = pd.concat([pv_ratio.rename(columns={"average": area}) for area in area_latlon.index], axis=1)

# Mapping of customers / identifiers to areas.
area_mapping = dict(zip(net_con.index, area_latlon.index, strict=True))
```

```python
# We still set the yearly average temperature
config = PVConfig(yearly_avg_temp=11.3, auto_infer_temp=False)

# Area-related data is provided at initialization
pv = PVDisaggregator(pv_ratio=pv_ratio_area, config=config,
                     area_mapping=area_mapping, area_latlon=area_latlon)

result = pv.disaggregate(net_con, net_gen)
```

Finally, instead of setting the an average temperature, you can also provide the temperature per time-point.

```python
# Provide temperature (in th)
temperature = pd.Series(
    # This just an example to demonstrate the functionality, provide real values here!
    [11.3] * len(pv_ratio.index),
    # Time points should be the same!
    index=pv_ratio.index,
)

# We still set the yearly average temperature
config = PVConfig(auto_infer_temp=False)

pv = PVDisaggregator(pv_ratio=pv_ratio, temperature=temperature, config=config)

result = pv.disaggregate(net_con, net_gen)
```
